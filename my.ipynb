```python
import numpy as np
np.random.seed(42)

A = np.random.randint(1, 11, size=(4, 3))
B = np.random.randint(1, 11, size=(3, 5))

print("Matrix A:")
print(A)
print("Shape of A:", A.shape)

print("\nMatrix B:")
print(B)
print("Shape of B:", B.shape)
assert A.shape[1] == B.shape[0], "Inner dimensions must match"

print("Dimension validation successful.")
print("A columns:", A.shape[1])
print("B rows:", B.shape[0])
C = np.dot(A, B)

print("Matrix C = A · B:")
print(C)

print("\nShape of C:", C.shape)

assert C.shape == (4, 5), "Resulting matrix must have shape (4, 5)"

print("Result shape verified successfully.")
C = A @ B
total_sum = np.sum(C)

print("Total sum of all elements in C:", total_sum)
row_sums = np.sum(C, axis=1)

print("Row-wise sums:")
print(row_sums)
column_sums = np.sum(C, axis=0)

print("Column-wise sums:")
print(column_sums)
```
